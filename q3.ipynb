```python
import pandas as pd 
import statsmodels.api as sm
import numpy as np 
import matplotlib.pyplot as plt 
```

(i) (5%) Select the sample from 1970-01-01 to 2015-12-01 and plot the sample.

```python
df = pd.read_csv("data/INDPRO.csv")    
start = np.where(df["observation_date"] =="1970-01-01")[0][0]
end = np.where(df["observation_date"] =="2015-12-01")[0][0]
df = df.iloc[start:end]
fig, ax =   plt.subplots()
plt.plot(df["observation_date"],df["INDPRO"], color  = "magenta")
plt.ylabel("INDPRO")
plt.xlabel("Observation Date")
plt.xticks(rotation = 45)
plt.xticks(plt.xticks()[0][::50])
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.show()


```
